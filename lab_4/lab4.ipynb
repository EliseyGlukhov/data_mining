```python
import pandas as pd

browser_df = pd.read_csv("data_browser.csv")
light_df = pd.read_csv("data_light.csv")

print(f"Браузер: {len(browser_df)} записей")
print(f"API:     {len(light_df)} записей")

browser_keys = set(zip(browser_df["text"], browser_df["author"]))
light_keys = set(zip(light_df["text"], light_df["author"]))

only_browser = browser_keys - light_keys
only_light = light_keys - browser_keys

print(f"Только в браузерном наборе: {len(only_browser)}")
print(f"Только в API-наборе:        {len(only_light)}")
print(f"Совпадение:                 {len(browser_keys & light_keys)}")
```
