# Задание №1

```python
import requests
import json
```

```python
USERNAME = 'EliseyGlukhov'
```

## Сохранение репозиториев в JSON файл

```python
response = (
    requests
    .get(f"https://api.github.com/users/{USERNAME}/repos")
    .json()
)

with open("repos.json", "w") as f:
    json.dump(response, f)
```

## Чтение репозиториев из JSON файла

```python
with open("repos.json", "r") as f:
    repos = json.load(f)
print(repos)
```

```python
print(f"All {USERNAME}'s repositories ({len(repos)} total):")
for i, repo in enumerate(repos):
    print(f"{i + 1}: {repo['name']}")
```

### Я взял API Яндекс диска, так как уже работал с ним

```python
request = requests.get('https://oauth.yandex.ru/authorize', params={
    'response_type': 'token', 
    'client_id': '863631e4fc014876a905961507c8cf50'
})
webbrowser.open(request.url)
# Токен берётся с появившейся страницы после авторизации, я его сохранил в JSON файл
with open("token.json", "r") as f:
    TOKEN = json.load(f)['token']
# Первые 10 символов токена
TOKEN[:10]
```

### Получаем все файлы на моём яндекс диске и сохраняем их названия в JSON файл

```python
path = f'disk:/'
files = requests.get('https://cloud-api.yandex.net/v1/disk/resources', params={
    'path': path
}, headers={
    'Authorization': TOKEN
}).json()  
result = []
for file in files['_embedded']['items']:
    result += [file['name']]

with open("disk_files.json", "w") as f:
    json.dump({'files': result}, f)
```

### Читаем созданный JSON файл

```python
with open("disk_files.json", "r") as f:
    disk_files = json.load(f)
print(disk_files)
```
